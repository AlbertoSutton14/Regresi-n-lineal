### Illustrating the Central Limit Theorem

The Central Limit Theorem states that the distribution of sample means of a sufficiently large number of independent, identically distributed random variables will be approximately normal, regardless of the underlying distribution of the original variables. Let's visualize this using the binomial distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameters for the binomial distribution (from previous cell, but adjusted for CLT illustration)
n_trials = 50 # Number of trials in each individual experiment
p_success = 0.5 # Probability of success

# Number of samples to draw
n_samples = 1000

# Size of each sample (number of observations in each sample mean calculation)
sample_size = 30 # A common rule of thumb for CLT is sample_size >= 30

# Generate multiple samples and calculate their means
sample_means = []
for _ in range(n_samples):
    # Generate a sample from the binomial distribution
    # np.random.binomial(n, p, size) generates 'size' number of outcomes for an (n, p) binomial distribution
    sample = np.random.binomial(n_trials, p_success, size=sample_size)
    sample_means.append(np.mean(sample))

# Plot the distribution of sample means
plt.figure(figsize=(10, 6))
sns.histplot(sample_means, kde=True, bins=30, color='skyblue')
plt.title('Distribution of Sample Means (Central Limit Theorem)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

As you can observe from the histogram above, even though the individual binomial distribution might not be perfectly normal (especially with small `n_trials`), the distribution of the *sample means* tends to be approximately normal. This is the essence of the Central Limit Theorem.
